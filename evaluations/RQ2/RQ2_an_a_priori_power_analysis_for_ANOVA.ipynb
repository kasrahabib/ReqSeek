```python
import numpy as np
from statsmodels.stats.power import FTestAnovaPower

alpha = 0.05
power = 0.80
effect_size = 0.50  # Cohen's f

k_groups = 5  # 0%, 10%, 20%, 30%, 40% inconsistency

analysis = FTestAnovaPower()

n_total = analysis.solve_power(effect_size = effect_size, alpha = alpha, power = power, k_groups = k_groups)

n_per_group = int(np.ceil(n_total / k_groups))

print("Required total sample size:", np.ceil(n_total))
print("Required balanced sample per group:", n_per_group)
print("Required balanced total:", n_per_group * k_groups)
print("Estimated power with 25 observations:", analysis.power(effect_size = effect_size, nobs = 25, alpha = alpha, k_groups = k_groups))
```
