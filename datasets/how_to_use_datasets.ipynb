## 5. Explore SwaRD

Inspect the dataset structure, available features, and number of instances. SwaRD classifies sentences into three categories using the `isReqSysAuxContAux` column: `requirement`, `system_related_auxiliary`, and `contextual_auxiliary`.

The following examples demonstrate how to filter requirement statements and display randomly selected instances from each class.

```python
# Overview: features (columns) and number of rows.
sward
```

```python
# SwaRD contains three classes in the "isReqSysAuxContAux" column:
# - "requirement": Identified software requirements.
# - "system_related_auxiliary": System-related non-requirement content.
# - "contextual_auxiliary": Contextual non-requirement content.
sward_only_requirements = sward.filter(lambda row: row["isReqSysAuxContAux"] == "requirement")
sward_only_requirements
```

```python
# Display up to five randomly selected examples from each SwaRD class.
import pandas as pd

sample_size = 5  # Maximum displayed examples per class.

def show_class_samples(name, dataset, class_column, seed):
    columns = ["REQID_expanded", "Project Name", "Requirement Sentences", class_column]
    missing = [column for column in columns if column not in dataset.column_names]
    if missing:
        raise KeyError(f"{name}: missing expected columns: {missing}")

    # Select just the displayed fields, then convert to pandas.
    # This DataFrame does NOT retain dataset.info licensing metadata.
    frame = dataset.select_columns(columns).to_pandas()
    counts = frame[class_column].value_counts(dropna = False).sort_index()
    display(Markdown(f"### {name} — Class Counts"))
    display(counts.rename_axis("Class").rename("Instances").to_frame())

    # Sample each category independently, using a fixed seed for repeatability.
    samples = pd.concat(
        [group.sample(n = min(sample_size, len(group)), random_state = seed)
         for _, group in frame.groupby(class_column, sort = True, dropna = False)],
        ignore_index=True,
    )
    
    display(Markdown(f"### {name} — Random Examples (up to {sample_size} per class)"))
    
    styled = (
        samples.style.hide(axis="index")
        .set_properties(**{
            "text-align": "left", "direction": "ltr",
            "font-size": "12px", "line-height": "1.2",
            "padding": "4px", "vertical-align": "top",
        })
        .set_table_styles([
            {"selector": "table", "props": [("table-layout", "fixed"), ("width", "100%") ]},
            {"selector": "th", "props": [("text-align", "left"), ("padding", "4px")]},
        ])
        .set_properties(subset=["Requirement Sentences"], **{
            "white-space": "normal", "word-break": "break-word", "max-width": "450px"
        })
        .set_properties(subset=["Project Name"], **{
            "white-space": "normal", "word-break": "break-word", "max-width": "200px"
        })
    )
    
    display(styled)

show_class_samples("SwaRD", sward, "isReqSysAuxContAux", seed = 42)
```

## 6. Convert Datasets to Other Formats

Hugging Face datasets support conversion to and from various data formats and structures, including pandas DataFrames, CSV, Parquet, JSON, and NumPy arrays, enabling integration with different data processing and machine learning workflows.

The following examples demonstrate three common conversions: **pandas, CSV, and Parquet**, along with previews of the converted data.

<p style="color: red;">
<strong>Note:</strong> Conversion may not preserve the original dataset metadata, including licensing and attribution information. Users are responsible for retaining the applicable information.
</p>

```python
import io
import pandas as pd

# Convert filtered SwaRD requirements to pandas.
sward_df = sward_only_requirements.to_pandas()

# Display five examples from pandas.
print("Pandas DataFrame:")
display(sward_df.head(5))
```

```python
# Convert to CSV in memory and display five examples.
csv_buffer = io.StringIO()
sward_df.to_csv(csv_buffer, index = False)
csv_buffer.seek(0)

print("CSV:")
display(pd.read_csv(csv_buffer).head(5))
```

```python
# Convert to Parquet in memory and display five examples.
parquet_buffer = io.BytesIO()
sward_df.to_parquet(parquet_buffer, index = False)
parquet_buffer.seek(0)

print("Parquet:")
display(pd.read_parquet(parquet_buffer).head(5))
```
